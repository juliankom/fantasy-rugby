```python
import pandas as pd
import os
dirpath = "rugbydata/player"
match_dirpath = "rugbydata/match"
```

```python
def select_matches(matches):
    accepted_competitions = set([
        'Six Nations', 
        'International Test Match', 
        'Internationals',
        'United Rugby Championship',
        'French Top 14',
        'Top 14',
        'Gallagher Prem',
        'Premiership',
    ])

    return matches[matches['competition'].apply(lambda c: c in accepted_competitions)]
```

```python
all_performances = pd.concat([pd.read_parquet(os.path.join(dirpath, f), engine="pyarrow") for f in os.listdir(dirpath) if "player_registry" not in f])
```

```python
all_matches = pd.concat([pd.read_parquet(os.path.join(match_dirpath, f), engine="pyarrow") for f in os.listdir(match_dirpath) if "match_registry" not in f])
```

```python
selected_matches = select_matches(all_matches)
```
