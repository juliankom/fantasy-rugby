```python
import pandas as pd
```

```python
matches = pd.read_parquet('selected_matches.parquet')
performances = pd.read_parquet('selected_performances.parquet')
```

```python
international_teams = set(['France', 'England', 'Ireland', 'Wales', 'Scotland', 'Italy'])
international_competitions = set(['International Test Match', 'Internationals'])
```

```python
matches = matches[
    ~ matches['competition'].isin(international_competitions) 
    | matches['home_team'].isin(international_teams) 
    | matches['away_team'].isin(international_teams)]
```

```python
performances = performances[performances['match_id'].isin(matches['match_id'])]
```
